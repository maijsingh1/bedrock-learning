```python
import boto3
import json

client_bedrock = boto3.client('bedrock-runtime')

model_id= "arn:aws:bedrock:us-east-1:429561518054:inference-profile/us.anthropic.claude-sonnet-4-6"

response = client_bedrock.invoke_model(
    modelId = model_id,
    body = json.dumps({
        "anthropic_version": "bedrock-2023-05-31",
        "max_tokens": 512,
        "messages": [
            {"role": "user", "content": "Hello! Can you summarize what this model is?"}
        ]
        })
)

result = json.loads(response["body"].read())
print(json.dumps(result, indent=2))

```
